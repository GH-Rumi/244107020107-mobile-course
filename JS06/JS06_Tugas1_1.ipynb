```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
# 1. Load dataset
df_voice = pd.read_csv('voice.csv')

# 2. Pisahkan fitur (X) dan target label (y)
X_voice = df_voice.drop('label', axis=1)
y_voice = df_voice['label']

# 3. Encode label (male = 1, female = 0)
le = LabelEncoder()
y_voice_encoded = le.fit_transform(y_voice)

# 4. Standardisasi fitur (Sangat penting untuk model SVM)
scaler = StandardScaler()
X_voice_scaled = scaler.fit_transform(X_voice)
```

```python
# Definisikan rasio test size (30% untuk 70:30, dan 20% untuk 80:20)
test_sizes = {'70:30': 0.3, '80:20': 0.2}

# Definisikan kernel yang akan digunakan
kernels = ['linear', 'poly', 'rbf']

# List untuk menyimpan hasil ke dalam tabel
results = []

# Langkah 4: Training dan Evaluasi
for split_name, test_size in test_sizes.items():
    # Split data
    X_train, X_test, y_train, y_test = train_test_split(
        X_voice_scaled, y_voice_encoded, test_size=test_size, random_state=42
    )

    for kernel in kernels:
        # Inisialisasi model SVM
        model = SVC(kernel=kernel)

        # Training model
        model.fit(X_train, y_train)

        # Prediksi
        y_pred = model.predict(X_test)

        # Hitung Akurasi
        acc = accuracy_score(y_test, y_pred)

        # Simpan hasil
        results.append({
            'Split Ratio': split_name,
            'Kernel': kernel,
            'Accuracy': acc
        })
```

```python
# Ubah list dictionary menjadi Pandas DataFrame agar rapi seperti tabel
df_results = pd.DataFrame(results)
print("=== Tabel Performansi Model SVM pada voice.csv ===")
print(df_results.to_string(index=False))
```
